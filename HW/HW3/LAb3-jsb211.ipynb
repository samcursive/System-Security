```python
#### Inputs
    # The function takes 3 inputs
    # string inputText: text to be encrypted
    # int N: number of positions to shift characters in the inputText by
    # Int D = direction of shift

#### Outputs
    # string encryptedText: encrypted version of inputTex
```

```python
# Algorithm
"""
1. Reverse the input text"
2. Once Shifted, all the ASCII characters in reversed input text by “N” positions.
3. 
"""
```

```python
# Reverse Function
def reverseInput(text):
    return "".join(reversed(text))
```

##### Step 1

```python
# example = "Sam"
# reversed_example = "".join(reversed(example))
# print(f"The Original String: {example}\nAnd the string Mirrored is: {reversed_example}")

MIN_ASCII_VAL = 34
MAX_ASCII_VAL = 126

inputText = input("Please enter a string: ")
#print(f"The User entered: {inputText}")
#reverse_input_Text = reverseInput(inputText)
#print(f"The Mirrored string is: {reverse_input_Text}")
```

##### Step 2

```python
def customEncrypt(inputText, N, D):
    if (N < 1):
        print( "Error: Cipher Key (N) must be greater than or equal to 1.")
    if (D != 1 and D != -1):
            print("Error: The Direction \'D\' must be either 1 (shift right) or -1 (shift left)")

    RANGE = MAX_ASCII_VAL - MIN_ASCII_VAL + 1 
    
    # Step 1: Reverse the input
    reverse_input_Text = reverseInput(inputText)
    
    # Take the reverse input and Apply Cipher to it. 
    def encrypt(character):
        encoding = ord(character)

        # Cannot have any spaces (32) and ! (33)
        if (encoding == 32 or encoding == 33):
                print("Illegal Character Detected: Space(s) or '!' are not allowed.")
                return character 
        
        # Shift
        shift = ((encoding - MIN_ASCII_VAL + (N*D)) % RANGE ) + MIN_ASCII_VAL
        
        return chr(shift)        

    # Mapping
    encryptedText = "".join(map(encrypt, reverse_input_Text))

    return encryptedText
```

```python
# Test function
customEncrypt("Sam", 2, -1)
```
